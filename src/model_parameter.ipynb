```python
from itertools import product

from model.cnn_bilstm import CNNBiLSTM 
```

```python
cnn_layers = [3, 5, 7]
rnn_layers = [1, 2, 3]


for cnn, rnn in product(cnn_layers, rnn_layers):
    model = CNNBiLSTM(num_classes=21, cnn_layers=cnn, rnn_layers=rnn)
    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)

    print(f"CNN layers: {cnn}, RNN layers: {rnn}")
    print(f"Parameters | Total : {total_params:,} | Trainable : {trainable_params:,}\n")
```
