```python
def ctc_greedy_decode_verbose(preds, idx2char, blank=0):
    print(f"Prediksi Mentah: {preds}")
    print(f"Decode Mentah: {[idx2char.get(int(p), f'ID:{int(p)}') for p in preds]}\n")

    decoded = []
    prev = blank
    
    for i, p in enumerate(preds):
        p = int(p)
        char = idx2char.get(p, f"ID:{p}")
        
        print(f"{i+1} Seq Preds '{char}' (Idx {p})")

        # LOGIKA CTC:
        # 1. Tidak boleh sama dengan karakter sebelumnya (menghapus duplikat berurutan)
        # 2. Tidak boleh karakter 'blank'
        if p != prev and p != blank:
            decoded.append(idx2char[p])
            print(f"   >> Simpan '{idx2char[p]}'")
        elif p == prev and p != blank:
            print(f"   >> Abaikan (Duplikat berurutan)")
        elif p == blank:
            print(f"   >> Abaikan (Karakter Blank)")
            
        prev = p

    hasil_akhir = "".join(decoded)
    print(f"--- Hasil Akhir: {hasil_akhir} ---\n")
    return hasil_akhir
```

```python
mapping = {0: "-", 1: "H", 2: "E", 3: "L", 4: "O"}

raw_output = [1, 1, 0, 2, 2, 2, 0, 3, 3, 0, 3, 4, 4, 4] 

hasil = ctc_greedy_decode_verbose(raw_output, mapping, blank=0)
```
