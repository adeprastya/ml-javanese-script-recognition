```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
# Forward pass

x_seq = [0.742, 0.136, 0.268, 0.461, 2.44]

W_f, U_f, b_f = 0.1, 0.1, 0.2
W_i, U_i, b_i = 0.2, 0.2, 0.1
W_o, U_o, b_o = 0.1, 0.2, 0.1
W_c, U_c, b_c = 0.2, 0.1, 0.2

h_prev = 0.0
C_prev = 0.0

print("=== LSTM Step by Step dengan Nilai Asli ===\n")

for t, x in enumerate(x_seq, 1):
    # Forget gate
    f = sigmoid(W_f * x + U_f * h_prev + b_f)
    # Input gate
    i = sigmoid(W_i * x + U_i * h_prev + b_i)
    # Candidate cell
    C_tilde = np.tanh(W_c * x + U_c * h_prev + b_c)
    # Cell state
    C = f * C_prev + i * C_tilde
    # Output gate
    o = sigmoid(W_o * x + U_o * h_prev + b_o)
    # Hidden state
    h = o * np.tanh(C)

    print(f"Timestep {t}")
    print(f"x_t = {x}")
    print(f"f_{t} = σ(W_f*x + U_f*h_prev + b_f) = {W_f}*{x} + {U_f}*{h_prev} + {b_f} = σ({W_f*x + U_f*h_prev + b_f}) =>  {f:.3f}")
    print(f"i_{t} = σ(W_i*x + U_i*h_prev + b_i) = {W_i}*{x} + {U_i}*{h_prev} + {b_i} = σ({W_i*x + U_i*h_prev + b_i}) => {i:.3f}")
    print(f"C̃_{t} = tanh(W_c*x + U_c*h_prev + b_c) = {W_c}*{x} + {U_c}*{h_prev} + {b_c} = tanh({W_c*x + U_c*h_prev + b_c}) => {C_tilde:.3f}")
    print(f"C_{t} = f*C_prev + i*C̃ = {f:.3f}*{C_prev:.3f} + {i:.3f}*{C_tilde:.3f}=  => {C:.3f}")
    print(f"o_{t} = σ(W_o*x + U_o*h_prev + b_o) = {W_o}*{x} + {U_o}*{h_prev} + {b_o} => {o:.3f}")
    print(f"h_{t} = o * tanh(C) = {o:.3f}*tanh({C:.3f}) = {0:.3f}*{np.tanh(C):.3f} => {h:.3f}\n")

    h_prev = h
    C_prev = C
```

```python
# Backward pass

x_seq = [0.742, 0.136, 0.268, 0.461, 2.44]

W_f, U_f, b_f = 0.1, 0.1, 0.2
W_i, U_i, b_i = 0.2, 0.2, 0.1
W_o, U_o, b_o = 0.1, 0.2, 0.1
W_c, U_c, b_c = 0.2, 0.1, 0.2

h_next = 0.0
C_next = 0.0

print("=== BiLSTM Backward Pass Step by Step ===\n")

for t_rev, x in enumerate(reversed(x_seq), 1):
    t = len(x_seq) - t_rev + 1  # timestep asli
    # Forget gate
    f = sigmoid(W_f * x + U_f * h_next + b_f)
    # Input gate
    i = sigmoid(W_i * x + U_i * h_next + b_i)
    # Candidate cell
    C_tilde = np.tanh(W_c * x + U_c * h_next + b_c)
    # Cell state
    C = f * C_next + i * C_tilde
    # Output gate
    o = sigmoid(W_o * x + U_o * h_next + b_o)
    # Hidden state
    h = o * np.tanh(C)

    print(f"Timestep {t} (backward)")
    print(f"x_t = {x}")
    print(f"f_{t} = σ(W_f*x + U_f*h_next + b_f) = {W_f}*{x} + {U_f}*{h_next} + {b_f} = σ({W_f*x + U_f*h_next + b_f}) =>  {f:.3f}")
    print(f"i_{t} = σ(W_i*x + U_i*h_next + b_i) = {W_i}*{x} + {U_i}*{h_next} + {b_i} = σ({W_i*x + U_i*h_next + b_i}) => {i:.3f}")
    print(f"C̃_{t} = tanh(W_c*x + U_c*h_next + b_c) = {W_c}*{x} + {U_c}*{h_next} + {b_c} = tanh({W_c*x + U_c*h_next + b_c}) => {C_tilde:.3f}")
    print(f"C_{t} = f*C_next + i*C̃ = {f:.3f}*{C_next:.3f} + {i:.3f}*{C_tilde:.3f} => {C:.3f}")
    print(f"o_{t} = σ(W_o*x + U_o*h_next + b_o) = {W_o}*{x} + {U_o}*{h_next} + {b_o} = σ({W_o*x + U_o*h_next + b_o}) => {o:.3f}")
    print(f"h_{t} = o * tanh(C) = {o:.3f}*tanh({C:.3f}) = {o:.3f}*{np.tanh(C):.3f} => {h:.3f}\n")

    h_next = h
    C_next = C
```
